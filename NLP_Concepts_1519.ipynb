NLP: Natural Language Processing
Two sections:
Natural Language Understanding(model understanding)
Natural Language Generation(model generates new content)
Text Understanding:
Text Cleaning:
converting the text content into lowercase
removal of punctuation marks, special characters
removal of stopwords

Stopwords:Give all the data to the model , this increases the complexity. To avoid that, we can remove the stopwords.

Words that are meaningful for sentence formation but not really necessary for our model to learn are referred as stop words.
Example: a,again,against,are,and ....

Tokenization:
Converting the words into characters representing tokens
Converting the sentences into words representing  tokens
Converting the paragraphs into sentences representing tokens

Example: ["beautiful","easy","easily","walked","walk","walking"] : total words machine has to learn - 6
[beauty,easy,easy,walk,walk,walk] - total words machine has to learn - 3

Stemming: Converts the word to its root form
It removes prefix/suffix so that root form of the word is resulted.
happiness : happy
given : give

3 most popular stemmer types:
Portal Stemmer
Lancaster Stemmer
Snowball Stemmer

Stemming techniques are resulting in a word that doesn't exist
happiness: happi
given : give
easily : easili


Lemmitization: Converts the word into base form. It uses the dictionary format. With the help of POS tags(parts of speech tags)
happiness: happiness
given: give

POS tags: Parts of Speech
Named Entity Recognition : (NER): Entities - you can image this as a set of entities/categories.
Oraganization
Location
Date/Time
Person


Example: "Sindhura works in 3RI technologies located at Pimple Saudagar. She has celebrated Ganesh Chaturthi on 27th August".
Sindhura,She - Person
3RI Technologies - Organization
Pimple Saudagar - Location
Date - 27th August
Vectorizer
Count Vectorizer(Bag of Words)

TFIDF

Word based vectors are generated that are provided to model for model training

# Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

words = ["easily","walking","happiness","given","marked","inspecting"]

ss = SnowballStemmer(language='english')

ss_words = [ss.stem(word) for word in words]
ss_words
```

```python
import spacy

# Load English pipeline
nlp = spacy.load("en_core_web_sm")

data = """
Data analysis is the process of inspecting, cleansing, transforming, and modeling data with the goal of discovering useful information, informing
conclusions, and supporting decision-making. Data analysis has multiple facets and approaches, encompassing diverse techniques under a
variety of names, and is used in different business, science, and social science domains.
In today's business world, data analysis plays a role in making decisions more scientific and helping businesses operate more effectively
"""

# Process text
doc = nlp(data)

# Extract chunks (noun phrases)
for chunk in doc.noun_chunks:
    print("NP:", chunk.text, " | Head:", chunk.root.text, " | Label:", chunk.root.dep_)
```

# **Deep Parsing**

```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

data = """
Data analysis is the process of inspecting, cleansing, transforming, and modeling data with the goal of discovering useful information, informing
conclusions, and supporting decision-making. Data analysis has multiple facets and approaches, encompassing diverse techniques under a
variety of names, and is used in different business, science, and social science domains.
In today's business world, data analysis plays a role in making decisions more scientific and helping businesses operate more effectively
"""

# process the data
doc = nlp(data)

# Print dependency relations
for token in doc:
    print(f"{token.text:<15} {token.dep_:<10} {token.head.text:<15} {token.pos_}")

# Visualize one sentence as dependency tree (in Jupyter)
displacy.render(list(doc.sents)[0], style="dep", jupyter=True, options={"distance": 120})
```

## **Named Entity Recognition**

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

# Sample data (keep capitalization)
data = """
Raman has graduated from California University of Education located at California on 24th August 2025.
"""

# Tokenize & POS tagging
word_tokens = word_tokenize(data)
pos_tags = pos_tag(word_tokens)

# Named Entity Recognition (chunking)
entities_tree = ne_chunk(pos_tags)

# Extract named entities
named_entities = []
for subtree in entities_tree:
    if hasattr(subtree, "label"):  # Check if it's a NE
        entity = " ".join([token for token, pos in subtree.leaves()])
        label = subtree.label()
        named_entities.append((entity, label))

print("Original Text:", data)
print("Named Entities:", named_entities)
```
